# IMDb Movies 1980 to 2020, What Drives Movie Revenue?

The goal of this project is to see which metrics correlate most with revenue and whether there are hints in the dataset about what drives the most revenue,
especially the budget/revenue ratio. It is important to note that gross revenue usually goes 50% to the cinemas and 50% to the movie studio, so the formula would be budget/(revenue/2). As a rule of thumb, gross box office revenue is split roughly 50/50 between cinemas and the studio (see [Marketplace, 2015](https://marketplace.org/2015/05/06/disney-vs-movie-theaters-next-box-office-fight)).

The dataset contains movies from all countries from the years 1980-2020. Column names are self explanetory, for example votes is number of votes, score is the weigthed average of the score, gross goes 50% to movie studios revenue and 50% to the cinema.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib
import matplotlib.pyplot as plt
import plotly.express as px
from io import BytesIO
import base64
from IPython.display import HTML

plt.style.use('ggplot')
matplotlib.rcParams['figure.figsize'] = (6, 4)
```

```python
# load data
# source: https://www.kaggle.com/datasets/danielgrijalvas/movies, IMDb
df = pd.read_csv(r'movies.csv')
```

```python
# production years of movies
print(f"Movies range from {df['year'].min():.0f} to {df['year'].max():.0f}")
```

```python
# columns
print(df.columns.tolist())
```

# Fix Columns

There is a mismatch between columns `year` and `released`. Let's assume that `released` is correct and that `year` might have been entered manually, with mistakes made. I will rename the `year` column to `false_year`, create a new `year` column from `released`, compare what percentage of rows have the same value, and then drop the `false_year` column.

```python
df = df.rename(columns={'year': 'false_year'})
df['year'] = df['released'].str.extract(r'(\d{4})').astype(float)
```

```python
# percent of matches
mismatch_pct = (df['year'] != df['false_year']).mean() * 100
print(f"{mismatch_pct:.0f}% of rows don't match in year and false_year")
```

```python
# drop false_year
df = df.drop(columns='false_year')
```

```python

# rename Unitet States to US and united kingdom to uk
df['country'] = df['country'].replace({'United States': 'US', 'United Kingdom': 'UK'})
```

# Revenue by Country

Lets see what countries have the most revenue cumulatively. Later I will adust for inflation, keep in mind infaltion from 1980 to 2020  is close to 400%. Lets list top 10 countries and then others in onecategory.

```python
gross_by_country = df.groupby('country')['gross'].sum().sort_values(ascending=False) / 1_000_000
top10 = gross_by_country.head(10)
top10['Other'] = gross_by_country.iloc[10:].sum()

tm = top10.reset_index()
tm.columns = ['country', 'gross']
print((top10 / 1000).map('{:,.1f}B'.format).rename_axis(None).to_string())
```

And now by percent of total

```python
total = gross_by_country.sum()

groups = {
    gross_by_country.index[0]: gross_by_country.iloc[0],
    gross_by_country.index[1]: gross_by_country.iloc[1],
    'Top 3-10': gross_by_country.iloc[2:10].sum(),
    'Other': gross_by_country.iloc[10:].sum(),
}

for name, value in groups.items():
    print(f"{name}: {value / total:.1%}")
```

```python
fig = px.treemap(tm, path=['country'], values='gross', title='Gross by country (millions)')
fig.update_traces(texttemplate='%{label}<br>%{value:,.0f}M<br>%{percentRoot:.1%}')
fig.show()
```

The US accounts for over 80% of total revenue. Next lets look at missing values and dtypes.

```python
# check dtype and missing values
pct = df.isnull().mean() * 100
n = len(pct)
colw = 90  # px per column

# chart exactly n * colw pixels wide, one point centered per column
fig = plt.figure(figsize=(n * colw / 100, 0.5), dpi=100)
ax = fig.add_axes([0, 0, 1, 1])
ax.plot(range(n), pct.values, marker='o', color='steelblue')
ax.set_xlim(-0.5, n - 0.5)
ax.set_ylim(-3, pct.max() + 3)
ax.axis('off')
buf = BytesIO()
fig.savefig(buf, format='png', transparent=True)
plt.close(fig)
img = base64.b64encode(buf.getvalue()).decode()

cell = f'style="width:{colw}px; padding:0; text-align:center"'
first = 'style="width:110px; padding:0; text-align:right"'

html = '<table style="table-layout:fixed; border-collapse:collapse">'
html += f'<tr><th {first}></th>' + ''.join(f'<th {cell}>{c}</th>' for c in pct.index) + '</tr>'
html += f'<tr><td {first}>dtype</td>' + ''.join(f'<td {cell}>{df[c].dtype}</td>' for c in pct.index) + '</tr>'
html += f'<tr><td {first}>missing</td>' + ''.join(f'<td {cell}>{p:.0f}%</td>' for p in pct) + '</tr>'
html += f'<tr><td {first}></td><td colspan="{n}" style="padding:0"><img src="data:image/png;base64,{img}" style="width:{n * colw}px; display:block"></td></tr>'
html += '</table>'

HTML(html)
```

The column buget has 28% missing values likely because some studios dont show their buget and keep it a secret. Yet besides gross it is one of most important columns so we cant drop it. Maybe the buget is only missing for smaller studios, lets see what corelation there is between missing buget and other columns, lets add a new column budget_missing with 0 and 1 for buget if missing.

 Before that, I will adjust for inflation to get a more accurate picture, using CPI data from FRED (Federal Reserve Economic Data).

```python
# check what does mising buget corelate most to? add a column for yes no mising buget
# flag: 1 if budget missing, 0 if not
df['budget_missing'] = df['budget'].isna().astype(int)

# flag: 1 if budget missing, 0 if not
df['budget_missing'] = df['budget'].isna().astype(int)

# pearson vs spearman correlation of the flag with numeric columns
corr = pd.DataFrame({
    'pearson': df.corr(numeric_only=True)['budget_missing'],
    'spearman': df.corr(numeric_only=True, method='spearman')['budget_missing'],
}).drop(['budget_missing', 'budget']).sort_values('spearman')

corr.round(2)
```

it looks like gross, votes and year are correlated, but number of votes itself is corelated with the buget and gross since biger movies have more ticket sold and votes. So lets plot year and gross. Most movies are over 10 mill and the have only 13% mising buget, and also 2020 seems a lot of buget is missing and there is very few movies in that year so we can exclude it also. Lets check what percent of movies is over 10mil

```python
# missing budget by year
missing_y = df.groupby('year')['budget'].apply(lambda x: x.isna().mean() * 100)
counts_y = df['year'].value_counts().sort_index()

# missing budget by gross
bins = [0, 1_000_000, 2_000_000, 5_000_000, 7_000_000, 10_000_000, np.inf]
labels = ['0-1M', '1-2M', '2-5M', '5-7M', '7-10M', '10M+']
df['gross_bin'] = pd.cut(df['gross'], bins=bins, labels=labels)
missing = df.groupby('gross_bin', observed=False)['budget'].apply(lambda x: x.isna().mean() * 100)
counts = df['gross_bin'].value_counts().sort_index()

fig, (ax3, ax1) = plt.subplots(1, 2, figsize=(12, 4))

# left: by year
ax1.bar(missing_y.index, missing_y.values)
ax1.set_title('Missing budget by year')
ax1.set_xlabel('Year')
ax1.set_ylabel('Missing budget (%)')
ax2 = ax1.twinx()
ax2.plot(counts_y.index, counts_y.values, color='black', marker='o', markersize=3)
ax2.set_ylabel('Number of movies')
ax2.set_ylim(bottom=0)
ax2.grid(False)

# right: by gross
missing.plot(kind='bar', ax=ax3)
ax3.set_title('Missing budget by gross')
ax3.set_xlabel('Gross')
ax3.set_ylabel('Missing budget (%)')
ax3.tick_params(axis='x', rotation=0)
ax3.text(5, missing['10M+'] + 1, f"{missing['10M+']:.0f}%", ha='center')
ax4 = ax3.twinx()
ax4.plot(range(len(counts)), counts.values, color='black', marker='o')
ax4.set_ylabel('Number of movies')
ax4.set_ylim(bottom=0)
ax4.grid(False)

plt.tight_layout()
plt.show()
```

```python
# what percent of movies has gross over 1 and 10 million?
for t in [1_000_000, 10_000_000]:
    pct = (df['gross'] > t).mean() * 100
    print(f"{pct:.0f}% of movies grossed over {t / 1_000_000:.0f} million")
```

62% of movues is over 10 million and for those ovies only 13% of them have mising buget. so lets drop movies with buget < 10mil and year = 2020.

It looks like the bigger gross the less missing buget, also most movies are in the category of over 10 milion. It is also much more interesting to big studios to look at movies with biger bugets. Fro these reasons we will drop movies under 10mil.

```python
# drop movies under 10mil and year isnt 2020
df = df[(df['gross'] >= 10_000_000) & (df['year'] != 2020)]
```

```python
# check duplicate rows
round(df.duplicated().mean() * 100, 2)
```

```python
# lets add estimated studio revenue, assuming ~50% of gross goes to cinemas
df['revenue_est'] = df['gross'] / 2
```

The dataset is mostly clean. We have no duplicates, missng values in buget dropped from 28 to 13%, and we have the movies only with the biggest gross over 10mil. Lets try now to build a predicttion chart for gross based on buget with linear regression.

For predictions lets drop the 13% mising buget

```python
clean = df.dropna(subset=['budget', 'gross'])
```

```python
# estimated studio revenue_est, ~50% of gross goes to cinemas
df['revenue_est'] = df['gross'] / 2

# only movies with both budget and revenue_est, in millions
clean = df.dropna(subset=['budget', 'revenue_est'])
x = clean['budget'] / 1e6
y = clean['revenue_est'] / 1e6

plt.figure(figsize=(6, 4))
sns.regplot(x=x, y=y, ci=None,
            scatter_kws={'color': 'red', 's': 10, 'alpha': 0.5},
            line_kws={'color': 'blue'})

plt.plot([0, x.max()], [0, x.max()], color='green', linestyle='--', label='buget = revenue_est')

slope, intercept = np.polyfit(x, y, 1)
corr = x.corr(y)
plt.title(f'Budget vs Revenue_est  |  r = {corr:.2f}  |  slope = {slope:.2f}')

# show
plt.xlim(0, x.max())
plt.ylim(0, y.max())
plt.xlabel('Budget (M)')
plt.ylabel('Revenue_est (M)')
plt.legend()
plt.show()
```

```python
# correlation heatmap
# hide 1s diagonal
corr = df.drop(columns=['budget_missing', 'revenue_est']).corr(numeric_only=True)
mask = np.triu(np.ones_like(corr, dtype=bool))

# drop empty first row and last column
corr = corr.iloc[1:, :-1]
mask = mask[1:, :-1]

plt.figure(figsize=(6, 4))
sns.heatmap(corr, mask=mask, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Correlation Matrix for Numeric Features')
plt.show()
```

```python
# top 10 regions by gross without the US, with budget
top10_country = (df.dropna(subset=['gross', 'budget'])
                   .query("country != 'US'")
                   .groupby('region')[['gross', 'budget']]
                   .sum()
                   .sort_values('gross', ascending=False)
                   .head(10))

(top10_country / 1e9).plot(kind='bar', figsize=(6, 4))
plt.title('Top 10 Countries by Gross (excluding US)')
plt.ylabel('Billion USD')
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
# top 10 movies by gross in each decade
d = df.dropna(subset=['year', 'gross']).copy()
d['decade'] = (d['year'] // 10 * 10).astype(int).astype(str) + 's'

top = d.sort_values('gross', ascending=False).groupby('decade').head(10).copy()
top['rank'] = top.groupby('decade').cumcount() + 1
top['movie'] = (top['name'] + ' (' + top['year'].astype(int).astype(str) + ', '
                + (top['gross'] / 1e6).round().astype(int).astype(str) + 'M)')

top.pivot(index='rank', columns='decade', values='movie')
```
